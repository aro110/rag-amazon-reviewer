# Dzień 4 — Zamiana recenzji na obiekty `Document`

Test funkcji `dataframe_to_documents` z `src/documents.py`.

```python
import sys
sys.path.append("..")

import pandas as pd
from src.documents import dataframe_to_documents

df = pd.read_csv("../data/processed/reviews_clean.csv")
documents = dataframe_to_documents(df)
```

## 1. Liczba dokumentów = liczba wierszy

```python
print(len(df), len(documents))
assert len(documents) == len(df)
```

## 2. Przykładowy dokument

```python
doc = documents[0]
print(doc.page_content)
print(doc.metadata)

assert doc.page_content == df.iloc[0]["Text"]
assert doc.metadata["ProductId"] == df.iloc[0]["ProductId"]
```

## 3. Brak pustych `page_content` i poprawne typy metadanych

```python
empty = [d for d in documents if not d.page_content.strip()]
print("Puste dokumenty:", len(empty))
assert len(empty) == 0

allowed_types = (str, int, float, bool, type(None))
for d in documents:
    for key, value in d.metadata.items():
        assert isinstance(value, allowed_types), (key, type(value))

print("Dokumenty z helpfulness_ratio = None:", sum(d.metadata["helpfulness_ratio"] is None for d in documents))
```
